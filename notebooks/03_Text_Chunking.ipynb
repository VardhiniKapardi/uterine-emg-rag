```python
from pathlib import Path
import pandas as pd

from langchain_text_splitters import RecursiveCharacterTextSplitter
```

```python
PROJECT_PATH = Path("/content/drive/MyDrive/uterine-emg-rag")

TEXT_PATH = PROJECT_PATH / "text"

print(TEXT_PATH)
```

```python
txt_files = sorted(TEXT_PATH.glob("*.txt"))

print(f"Found {len(txt_files)} text files\n")

for file in txt_files:
    print(file.name)
```

```python
sample_file = txt_files[0]

with open(sample_file, "r", encoding="utf-8") as f:
    text = f.read()

print(text[:1000])
```

```python
print(len(text), "characters")
print(len(text.split()), "words")
```

```python
splitter = RecursiveCharacterTextSplitter(
    chunk_size=1000,
    chunk_overlap=200
)
```

```python
chunks = splitter.split_text(text)
```

```python
print(len(chunks))
```

```python
print(chunks[0])
```

```python
print(chunks[1])
```

```python
all_chunks = []
```

```python
for file in txt_files:

    with open(file, "r", encoding="utf-8") as f:
        text = f.read()

    chunks = splitter.split_text(text)

    for i, chunk in enumerate(chunks):

        all_chunks.append({

            "paper": file.stem,

            "chunk_id": i,

            "text": chunk

        })
```

```python
print(len(all_chunks))
```

```python
all_chunks[0]
```

```python
chunks_df = pd.DataFrame(all_chunks)

chunks_df.head()
```

```python
chunks_df.groupby("paper").size()
```

```python
chunks_df["characters"] = chunks_df["text"].apply(len)

chunks_df["characters"].describe()
```

```python
OUTPUT_PATH = PROJECT_PATH / "processed"

OUTPUT_PATH.mkdir(exist_ok=True)

chunks_df.to_csv(
    OUTPUT_PATH / "chunks.csv",
    index=False
)
```

```python
chunks_df.to_json(
    OUTPUT_PATH / "chunks.json",
    orient="records",
    indent=2
)
```
