```python
import os

from langchain_community.document_loaders import PyMuPDFLoader
```

```python
PAPER_PATH = "/content/drive/MyDrive/uterine-emg-rag/papers"
```

```python
pdf_files = []

for file in os.listdir(PAPER_PATH):
    if file.endswith(".pdf"):
        pdf_files.append(file)

print(pdf_files)
```

```python
sample_pdf = os.path.join(PAPER_PATH, pdf_files[0])

loader = PyMuPDFLoader(sample_pdf)

documents = loader.load()
```

```python
print(len(documents))
```

```python
print(documents[0].page_content[:1500])
```

```python
documents[0].metadata
```

```python
all_documents = []

for pdf in pdf_files:

    loader = PyMuPDFLoader(
        os.path.join(PAPER_PATH, pdf)
    )

    docs = loader.load()

    all_documents.extend(docs)
```

```python
print(len(all_documents))
```

```python
for doc in all_documents[:5]:
    print(doc.metadata)
```
