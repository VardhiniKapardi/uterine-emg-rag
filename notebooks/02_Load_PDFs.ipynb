```python
import os
from pathlib import Path
import fitz  # PyMuPDF
```

```python
PROJECT_PATH = Path("/content/drive/MyDrive/uterine-emg-rag")

PAPERS_PATH = PROJECT_PATH / "papers"

print(PAPERS_PATH)
```

```python
print(PAPERS_PATH.exists())
```

```python
pdf_files = sorted(PAPERS_PATH.glob("*.pdf"))

print(f"Found {len(pdf_files)} PDFs\n")

for pdf in pdf_files:
    print(pdf.name)
```

```python
def extract_text(pdf_path):

    document = fitz.open(pdf_path)

    text = ""

    for page in document:

        text += page.get_text()

    document.close()

    return text
```

```python
sample_text = extract_text(pdf_files[0])

print(sample_text[:1500])
```

```python
len(sample_text.split())
```

```python
papers = {}
```

```python
for pdf in pdf_files:

    text = extract_text(pdf)

    papers[pdf.stem] = text
```

```python
print(len(papers))
```

```python
papers.keys()
```

```python
print(papers["paper1"][:1500])
```

```python
for name, text in papers.items():

    words = len(text.split())

    print(f"{name}: {words} words")
```

```python
summary = []

for name, text in papers.items():

    summary.append({

        "Paper": name,

        "Words": len(text.split())

    })
```

```python
import pandas as pd

summary_df = pd.DataFrame(summary)

summary_df
```

```python
summary_df.sort_values("Words", ascending=False)
```

```python
summary_df["Words"].sum()
```

```python
TEXT_PATH = PROJECT_PATH / "text"

TEXT_PATH.mkdir(exist_ok=True)
```

```python
for name, text in papers.items():

    with open(TEXT_PATH / f"{name}.txt", "w", encoding="utf-8") as f:

        f.write(text)
```
